```python
from sklearn import tree
```

```python
x = [[3, 3], [4, 3]]
y = [23, 34]
```

```python
clf = tree.DecisionTreeRegressor(max_depth=1,max_features=1)
clf.fit(x, y)
```

```python
v = clf.predict([[3,3]])
print(v)
import random
list=[]
for x in range(2):
    list1=[]
    x=random.randint(1,6)
    y=random.randint(1,6)
    list1.append(x)
    list1.append(y)
    list.append(list1)


pre = []
for x in list:
    k = clf.predict([x])
    pre.append(k[0])  
print(pre)

```

```python
from sklearn import  metrics
print(y)
print(metrics.confusion_matrix(y,pre))

```
